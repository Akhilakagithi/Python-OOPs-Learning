```python
#with constructor
class Student:
    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course

s1 = Student("Rahul", 22, "Python")#each object can have different attribute values
s2 = Student("Priya", 21, "Java")

print(s1.name, s1.age, s1.course)
print(s2.name, s2.age, s2.course)
```

## Task 4: Calculate Area of Rectangle

**Concepts:** Class, Object, Constructor and Methods.

### Requirements
1. Create a class called Rectangle.
2. Use a constructor.
3. Accept length and breadth.
4. Create a calculate_area() method.
5. Create two objects and calculate their areas.

### Expected Output
Rectangle 1: Area = 50
Rectangle 2: Area = 32


```python
class Rectangle():
    def __init__(self,rectangle_id,length,breadth):
        self.rectangle_id=rectangle_id
        self.length=length
        self.breadth=breadth

    def calculate_area(self):#method:it is a functio defined inside class it describes an action that an object can perform
        area=self.length*self.breadth
        print("Rectangle",self.rectangle_id,)
        print("Length:",self.length)
        print("Breadth:",self.breadth)
        print("Area",area)
        print(" ")

r1=Rectangle(1,10,5)
r2=Rectangle(2,8,4)

r1.calculate_area()
r2.calculate_area()
```

Problem statement
Create a Python class called Student to calculate the total marks, average marks, and grade of a student.
Requirements
1. Create a class called Student.
2. Create a constructor accepting:
   - student_id
   - name
   - mark1
   - mark2
   - mark3
3. Create a method called calculate_total() to calculate total marks.
4. Create a method called calculate_average() to calculate average marks.
5. Create a method called display_result() to display student details, total, average, and grade.
6. Create two student objects with different marks.
Grade conditions
Average marks	Grade
90 and above	A
75–89	B
60–74	C
40–59	D
Below 40	Fail

Sample student details
- Student 1: Rahul, 85, 90, 95
- Student 2: Priya, 70, 75, 80
Expected output
Student 1 Details:
Name: Rahul
Total Marks: 270
Average: 90.0
Grade: A

Student 2 Details:
Name: Priya
Total Marks: 225
Average: 75.0
Grade: B

```python
class Student():
    def __init__(self,student_id,name,mark1,mark2,mark3):
        self.student_id=student_id
        self.name=name
        self.mark1=mark1
        self.mark2=mark2
        self.mark3=mark3

    def calculate_total(self):
        print("student",self.student_id,"Details")
        print("Name:",self.name)
        self.total_marks=self.mark1+self.mark2+self.mark3
        print("Total Marks:",self.total_marks)

    def calculate_average(self):
        self.average=self.total_marks/3
        print("Average:",self.average)
        
    def display_result(self):
        if self.average>=90:
            print("Grade: A")
        elif self.average>=75 and self.average<=89:
            print("Grade: B")
        elif self.average >=60 and self.average <=74:
            print("Grade: C")
        elif self.average >=40 and self.average <=59:
            print("Grade: D")
        else:
            print("Fail")
        print()
            
s1=Student(1,"Rahul",85,90,95)
s2=Student(2,"Priya",70,75,80)
s1.calculate_total()
s1.calculate_average()
s1.display_result()

s2.calculate_total()
s2.calculate_average()
s2.display_result()
```
