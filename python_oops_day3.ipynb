Task: Student Marks Management using Getter and Setter

Problem statement:

Create a Student class to manage student marks using getter and setter methods.

Requirements:
1. Create a class called Student.
2. Create a constructor accepting name and marks.
3. Store marks using a private-style attribute __marks.
4. Create a getter method get_marks() to retrieve marks.
5. Create a setter method set_marks(new_marks):
   - If marks are between 0 and 100, update the marks.
   - Otherwise, print "Invalid marks".
Sample details:
- Name: Rahul
- Initial marks: 75
- New marks: 90
- Invalid marks: 150
Expected output:
Student Name: Rahul
Current Marks: 75
Marks updated successfully
Updated Marks: 90
Invalid marks

```python
class Student():
    def __init__ (self,name,marks):
        self.name=name
        self.__marks=marks

    def get_marks(self):
        return self.__marks

    def set_marks(self,new_marks):
        if new_marks>0 and new_marks<100:
            self.__marks=new_marks
            print("Marks updated successfully")
            print("Updated Marks:",self.__marks)
        else:
            print("Invalid Marks")

    def display_info(self):
        print("Student Name:",self.name)
        print("Current Marks:",self.__marks)

s=Student("Rahul",75)
s.display_info()
#s.get_marks()
s.set_marks(90)
s.set_marks(150)
#print("updated m:",s.get_marks())



            
```
