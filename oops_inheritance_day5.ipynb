#method overriding
Create a parent class Vehicle and a child class Car.

Requirements:
1. Create a parent class Vehicle.
2. Add a method start() that prints "Vehicle is starting".
3. Create a child class Car that inherits from Vehicle.
4. Override the start() method inside Car to print "Car is starting with a button".
5. Create a Car object and call start().
Expected output:
Car is starting with a button

```python
class Vehicle(): #parent class
    def start(self):    #same method name
        print("Vehicle is starting")

class Car(Vehicle):#child class
    def start(self):#same method name
        
        print("Car is starting with a button")

c=Car()
c.start()#python executes child's method
```

# Can we still call the parent class's start() method from inside the child class after overriding it?
-> Yes, using super().start().

```python
class Vehicle(): #parent class
    def start(self):    #same method name
        print("Vehicle is starting")

class Car(Vehicle):#child class
    def start(self):#same method name
        super().start()
        
        print("Car is starting with a button")

c=Car()
c.start()#python executes child's method
```

Advantage = Method overriding allows different child classes to use the same method name but perform different actions according to their requirements.
