Task 6: Bank Account Management
Encapsulation + Class + Object + Constructor + Methods
Problem statement
Create a Python class called BankAccount to manage a customer's bank balance.
Requirements
1. Create a class called BankAccount.
2. Create a constructor accepting account_holder and balance.
3. Store balance as a private attribute using __balance.
4. Create a method called deposit(amount) to add money.
5. Create a method called withdraw(amount) to withdraw money.
6. Create a method called show_balance() to display the current balance.
7. Add validation:
   - Deposit amount must be greater than zero.
   - Withdrawal amount must be greater than zero.
   - Withdrawal amount must not exceed the available balance.
8. Create one bank account object and test all three methods.
Sample details
Details	Value
Account holder	Rahul
Initial balance	10,000
Deposit amount	2,000
Withdrawal amount	3,000

Expected output
Account Holder: Rahul
Initial Balance: 10000

Amount deposited successfully.
Current Balance: 12000

Amount withdrawn successfully.
Current Balance: 9000

```python
class BankAccount:
    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.__balance=balance

    def deposit(self,amount):
        if amount>0:
            #self.__balance=self.__balance+amount
            self.__balance+=amount
            print("Amount deposited successfully.")
            print("Current Balance:",self.__balance)
            print()

    def withdraw(self,amount):
        if amount>0 and amount<=self.__balance:
            self.__balance-=amount
            print("Amount withdrawn successfully.")
            print("Current Balance:",self.__balance)

    def show_balance(self):
        print("Account Holder:",self.account_holder)
        print("Initial_Balance:",self.__balance)
        print()

o=BankAccount("Rahul",10000)
o.show_balance()
o.deposit(2000)
o.withdraw(3000)
```
