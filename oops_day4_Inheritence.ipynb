concept of inheritence

Task: Animal and Dog

Requirements:
1. Create a parent class called Animal.
2. Create a method eat() that prints "Animal is eating".
3. Create a child class called Dog that inherits from Animal.
4. Create a method bark() that prints "Dog is barking".
5. Create an object of Dog.
6. Call both eat() and bark() methods.
   
Expected output:
Animal is eating
Dog is barking

```python
class Animal():#parent class
    def eat(self):
        print("Animal is eating")

class Dog(Animal):#child class and aquaring properties from parent class
    def bark(self):
        print("Dog is barking")

d=Dog()#object created for child class
d.eat()
d.bark()

```

```python
class Animal():#parent class
    def eat(self):
        print("Animal is eating")

class Dog(Animal):#child class and aquaring properties from parent class
    def bark(self):
        print("Dog is barking")
a=Animal()# object created for parent class
a.eat()
a.bark()
```

Your practice task – Task 9
Create a parent class Person and a child class Student.

Requirements:
1. Person should have a constructor that accepts name and age.
2. Student should inherit from Person.
3. Student should have its own constructor accepting name, age, and course.
4. Use super() to initialize name and age.
5. Initialize course inside the Student constructor.
6. Create one Student object and display all three details.
Sample object:
s1 = Student("Rahul", 22, "Python")



Expected output:
Name: Rahul
Age: 22
Course: Python



```python
class Person():
    def __init__(self,name,age):
        self.name=name
        self.age=age
        
class Student(Person):
    def __init__(self,name,age,course):
        super().__init__(name,age)
        #super().__init__(age)
        self.course=course

s=Student("Rahul",22,"Python")
print("Name:",s.name)
print("Age:",s.age)
print("Course:",s.course)
    
```
